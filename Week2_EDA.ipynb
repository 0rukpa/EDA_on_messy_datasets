```python
df.drop_duplicates(inplace=True)
```

```python
df.drop('Cabin',inplace=True,axis=1)
```

```python
#changing my ticket column to str as  i made an error in converting to numeric first
df['Ticket'] = df['Ticket'].astype(str)

#using regex
df['Ticket'] = df['Ticket'].str.replace(r'\D', '', regex=True)

df['Ticket'] = pd.to_numeric(df['Ticket'], errors='coerce')

df['Ticket'] = df['Ticket'].astype('Int64')

print(df['Ticket'].head())
```

```python
df.head(5)
```

```python
#filling up age's missing values with its mean
df['Age'] = df['Age'].fillna(df['Age'].mean())
print("Succesfully filled up Age with mean")
```

```python
df['Embarked']= df['Embarked'].fillna(df['Embarked'].mode())
print("Succesfully completed the 'Embarked' Column")
```

```python
survivors=len(df[df['Survived']==1])*100/len(df)
victim=len(df[df['Survived']==0])*100/len(df)

print(f'Percentage of passengers who survived:', survivors)
print(f'Percentage of passengers who did not survive:', victim)
```
